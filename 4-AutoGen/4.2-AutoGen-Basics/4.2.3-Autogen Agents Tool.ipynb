# Tools

Tools are code that can be executed by an agent to perform actions. A tool can be a simple function such as a calculator, or an API call to a third-party service such as stock price lookup or weather forecast. In the context of AI agents, tools are designed to be executed by agents in response to model-generated function calls.


There are three types of Tools :<br>
1. Custom Function Tools
2. Built-in Tools
3. Third Party Calling Tools (eg. Langchain built-in tools)

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_core.tools import FunctionTool
```

```python
import asyncio, os

from dotenv import load_dotenv

# Load environment variables
load_dotenv()
api_key= os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("Please set the OPENAI_API_KEY environment variable.")
```

```python
model_client= OpenAIChatCompletionClient(model="gpt-4o", api_key=api_key)
```

```python
def reverse_string(text: str) -> str:
    '''
    Reverse the given text

    input: str

    output: str

    The reverse string is returned.
    '''
    return text[::-1]
```

```python
# We can pass only one custom function within FunctionTool at a time
reverse_tool= FunctionTool(reverse_string, description="A tool to reverse a string")
```

```python
reverse_tool
```

```python
agent= AssistantAgent(
    name="ReverseStringAgent",
    model_client=model_client,
    system_message='You are a helpful assistant that can reverse string using reverse_string tool. Give the result with summary',
    tools=[reverse_tool] # Here, We can pass multiple tools
)
```

```python
agent
```

```python
response= await agent.run(task='"Reverse the string "Hello, World!"')
```
