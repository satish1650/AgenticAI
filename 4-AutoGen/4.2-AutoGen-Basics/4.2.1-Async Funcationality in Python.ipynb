### Synchronous or Normaly way of Coding 

```python
import time

def brew_coffee():
    print("Brewing Coffee")
    time.sleep(3) # 3 minutes
    print("Coffee Ready")

def toast_bagel():
    print("Toasting bagel")
    time.sleep(2) # 2 minutes
    print("Bagel Ready")

def main():
    start= time.time()

    coffee= brew_coffee()
    bagel= toast_bagel()

    end= time.time()
    print(f"Time:{end-start:.2f} minutes")

main()
```

### Asynchronous Version

```python
import time
import asyncio # Asynchronous Input Output Library

async def brew_coffee():
    print("Brewing Coffee")
    await asyncio.sleep(3) # 3 minutes
    print("Coffee Ready")

async def toast_bagel():
    print("Toasting bagel")
    await asyncio.sleep(2) # 2 minutes
    print("Bagel Ready")

async def main():
    start= time.time()

    coffee= brew_coffee()
    bagel= toast_bagel()

    results= await asyncio.gather(coffee, bagel)

    end= time.time()
    print(f"Time:{end-start:.2f} minutes")


await main()
```
