# Agentchat API with Assistant Agent

```python
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
api_key= os.getenv("OPENAI_API_KEY")
```

```python
# Define Model -> Brain

model_client= OpenAIChatCompletionClient(model="gpt-4o", api_key=api_key)
```

```python
# Define Assistant -> Digital Human -> Name and Brain

my_assistant= AssistantAgent(name="Assistant", model_client=model_client)
```

```python
response= my_assistant.run(task="Who are you?")
```

```python
response # Not getting result because not used await here
```

```python
# run is an async method so we have to use await 

response= await my_assistant.run(task="Who are you?")
```
