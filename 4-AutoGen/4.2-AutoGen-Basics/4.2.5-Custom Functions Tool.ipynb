# Custom Function Tools

A tool can also be a simple Python function that performs a specific action. To create a custom function tool, you just need to create a Python function and use the **`FunctionTool`** class to wrap it.

The **`FunctionTool`** class uses descriptions and type annotations to inform the LLM when and how to use a given function. The description provides context about the function’s purpose and intended use cases, while type annotations inform the LLM about the expected parameters and return type.

### Below Example of Custom Function Tools

```python
import asyncio, os
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_core.tools import FunctionTool
from dotenv import load_dotenv
```

```python
# Load environment variables
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError("Please set the OPENAI_API_KEY environment variable.")
```

```python
model_client= OpenAIChatCompletionClient(model="gpt-4o", api_key=api_key)
```

```python
model_client
```

```python
def reverse_string(text: str) -> str:
    '''
    Reverse the given text

    input: str

    output: str

    The reverse string is returned.
    '''
    # return "Hello how are you?"
    return text[::-1]
```

```python
reverse_tool= FunctionTool(reverse_string, description="A tool to reverse a string")
```

```python
reverse_tool
```

```python
agent= AssistantAgent(
    name="ReverseStringAgent",
    model_client=model_client,
    system_message="You are a helpful assistant that can reverse string using reverse_string tool. Give the result with summary",
    tools=[reverse_tool],
    reflect_on_tool_use=False
)
```

```python
response= await agent.run(task='Reverse the string "Hello, World!"')
```
