# **AutoGen Studio**

`AutoGen Studio` is a low-code interface built to help you rapidly prototype AI agents, enhance them with tools, compose them into teams and interact with them to accomplish tasks. It is built on AutoGen AgentChat - a high-level API for building multi-agent applications.

See a video tutorial on AutoGen Studio v0.4 (02/25) - https://youtu.be/oum6EI7wohM

See one more video tutorial on AutoGen Studio - https://www.youtube.com/watch?v=Z0J4Hdo_d78

Code for AutoGen Studio is on GitHub at microsoft/autogen (https://github.com/microsoft/autogen/tree/main/python/packages/autogen-studio)

***Reference URL -*** https://microsoft.github.io/autogen/stable//user-guide/autogenstudio-user-guide/index.html

`Please install the below latest autogen studio package (autogenstudio 0.4.2.1)`

https://pypi.org/project/autogenstudio/0.4.2.1/

The below setup is required for autogen Studio
1. Create a new environemet:- <br>
    > python -m venv autogen-studio-venv <br>
    > conda create -p autogen-studio-venv <br>
    > conda create -p autogen-studio-venv python==3.12.11 -y

2. Activate the environment <br>
    > source autogen-studio-venv/bin/activate <br>
    > conda activate autogen-studio-venv/

3. Install the autogen studio (this is the latest one till 19th July 2025) <br>
    > pip install autogenstudio==0.4.2.1 <br>

4. Export your OPENAI_API_KEY <br>

5. Run the application <br>
    > autogenstudio ui <br>

After completed above all 5 steps, we will have to run `autogenstudio ui` then below display will come, will have to copy http://127.0.0.1:8081 url and paste to new window's browser



> Application startup complete. Navigate to http://127.0.0.1:8081
