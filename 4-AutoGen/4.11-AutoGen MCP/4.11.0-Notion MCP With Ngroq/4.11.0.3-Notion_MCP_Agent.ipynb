### `Cross-Origin Resource Sharing (CORS)`
<hr>

`CORS`, or **Cross-Origin Resource Sharing**, is a security feature implemented by web browsers that controls access to resources on different domains (origins). It allows a web page to request resources from a server on a different domain, but only if the server explicitly allows it through HTTP headers. This mechanism helps prevent malicious websites from accessing sensitive data from other sites without permission. 
<hr>

##### References
https://developer.mozilla.org/en-US/docs/Web/HTTP/Guides/CORS
